```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

matricula = pd.read_csv("matricula_sistemas.csv")
consumo = pd.read_csv("consumo_laboratorios.csv")

print("Datos de matrícula:")
display(matricula.head())

print("Datos de consumo:")
display(consumo.head())
```

```python
matricula.describe()
```

```python
col = matricula["num_matriculados"]

print("Media:", col.mean())
print("Mediana:", col.median())
print("Moda:", col.mode().tolist())
print("Rango:", col.max() - col.min())
print("Varianza:", col.var())
print("Desviación estándar:", col.std())
```

La matrícula tiene un promedio de 137.08 estudiantes por semestre y una mediana de 137. La moda es múltiple: 120, 137 y 160 son los valores que más se repiten. El rango es de 123 estudiantes, lo que muestra una diferencia considerable entre el mínimo y el máximo. La desviación estándar es de aproximadamente 34.12, por lo que la matrícula suele variar alrededor de 34 estudiantes respecto al promedio.
